```python
import os, time, csv
from datetime import datetime, timezone
from openai import OpenAI
from kaggle_secrets import UserSecretsClient

MODEL_NAME, GROQ_ID = "gpt-oss-120b", "openai/gpt-oss-120b"
TEMPERATURE, MAX_TOKENS = 0.0, 512
OUT = "/kaggle/working/gptoss_missing22.csv"
COLS = ["model","strategy","question_id","domain","trap_type","orig_category","prompt","response",
        "finish_reason","prompt_tokens","completion_tokens","seconds","timestamp_utc"]

key = UserSecretsClient().get_secret("GROQ_API_KEY")
assert key, "Add GROQ_API_KEY in Add-ons > Secrets and switch it on for this notebook"
client = OpenAI(base_url="https://api.groq.com/openai/v1", api_key=key)

done = set()
if os.path.exists(OUT):
    with open(OUT, newline="", encoding="utf-8") as f:
        for r in csv.DictReader(f):
            if (r["response"] or "").strip():
                done.add((r["strategy"], int(r["question_id"])))
print("already done:", len(done))

def ask(prompt):
    extra = {"reasoning_effort": "low", "include_reasoning": False}
    for attempt in range(8):
        try:
            t0 = time.time()
            r = client.chat.completions.create(model=GROQ_ID, messages=[{"role": "user", "content": prompt}],
                    temperature=TEMPERATURE, max_completion_tokens=MAX_TOKENS, extra_body=extra)
            u = r.usage
            return ((r.choices[0].message.content or "").strip(), r.choices[0].finish_reason,
                    u.prompt_tokens if u else 0, u.completion_tokens if u else 0, round(time.time()-t0, 2))
        except Exception as e:
            msg = str(e)
            if "per day" in msg or "(RPD)" in msg or "(TPD)" in msg:
                raise RuntimeError("Groq DAILY limit reached. Try again after the limit resets; finished rows are kept.")
            if "401" in msg or "invalid_api_key" in msg:
                raise RuntimeError("Groq key rejected. Check the GROQ_API_KEY secret.")
            wait = min(60, 5 * 2 ** attempt)
            print(f"   retry {attempt+1} in {wait}s: {msg[:150]}")
            time.sleep(wait)
    raise RuntimeError("Groq call failed 8 times")

# keep only the latest row per prompt when resuming
rows = []
if os.path.exists(OUT):
    with open(OUT, newline="", encoding="utf-8") as f:
        rows = [r for r in csv.DictReader(f) if (r["strategy"], int(r["question_id"])) in done]

for j in JOBS:
    k = (j["strategy"], j["question_id"])
    if k in done: continue
    text, fin, pt, ct, sec = ask(j["prompt"])
    row = {"model": MODEL_NAME, **{c: j[c] for c in ["strategy","question_id","domain","trap_type","orig_category","prompt"]},
           "response": text, "finish_reason": fin, "prompt_tokens": pt, "completion_tokens": ct, "seconds": sec,
           "timestamp_utc": datetime.now(timezone.utc).strftime("%Y-%m-%dT%H:%M:%S")}
    rows.append(row)
    with open(OUT, "w", newline="", encoding="utf-8") as f:
        w = csv.DictWriter(f, fieldnames=COLS); w.writeheader(); w.writerows(rows)
    print(f"{j['strategy']:12s} Q{j['question_id']:<3d} {fin:6s} {ct:4d} tok | {text[:70]!r}")
    time.sleep(8)   # stays under the 8k tokens-per-minute free-tier limit
```
